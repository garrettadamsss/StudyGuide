### MergeSort
1. Divide: Recursively split the array into two halves until left with single elements
2. Conquer: Merge each halve into a single sorted array 

```python
def merge(left: list[int], right: list[int]): 
    # merge left and right arrays
    # compare current value of each and merge the smaller value
    i = 0
    j = 0
    result = []

    while i < len(left) and j < len(right):
        if(left[i] <= right[j]): 
            result.append(left[i])
            i += 1
        else: 
            result.append(right[j])
            j += 1

    # append any remaining items from both arrays
    while i < len(left):
        result.append(left[i])
        i += 1
    while j < len(right):
        result.append(right[j])
        j += 1
    return result; 



def merge_sort(arr: list[int]):
    # 1. divide
    # base case
    if (len(arr) <= 1):
        return arr; 

    mid = len(arr) // 2

    left = merge_sort(arr[0:mid]);
    right = merge_sort(arr[mid:len(arr)]);

    # 2. merge
    return merge(left, right);

arr = [1, 5, 4, 3, 2];

print(merge_sort(arr))
```

# Recursion

Think of recursion as a stack of function calls where each function call is popped and returns a value to the previous function. Each function call is independent from the others, the arguement is not mutated. Think about it like a while true loop with break clause? 

You are solving a bigger problem with sub problems. For example If I want to find 5-th member of the fibonacci sequence, I need to know 4-th and 3-rd. But to know 4-th I need 3-rd and 2-nd and so on. You find the solution to a big problem by first finding a solution for the smaller problem. So first write a return statement that solves that problem with smaller problems 

In case of Fibonacci, as mentioned, for any n you need to know n-1 and n-2.
Therefore, if we want to find fib(n)  we need to write:
return fib(n-1) + fib(n-2).

Now we need a base case which is an answer to a problem we already know. 
We know fib(1) && fib(2) = 1, so we write:
if(n <= 2) {
    return 1; 
}
Components of a recursive function: 
1. Define a base case: Identify the simplest/base case where the solution is known. This is the stopping condition that prevents the function from infinitely calling itself. 
2. Define a recursive case: Define the problem in terms of smaller problems. Break down the problem into a smaller subproblems and call the function to recursively solve each subproblem 
3. Combine the solutions of the subproblems into a single solution 


```python
# Fib
# each succeeding number is the sum of the two previous numbers

def fib(n: int):
    # 1. Define base case
    if n == 0:
        return 0
    if n == 1:
        return 1
    # 2. Break down into subproblems
    # 3. Return the combined solution 
    return fib(n - 2) + fib(n - 1)

print(fib(10))
```

# Binary Search
1. Start with low index(0) and high index(len(list) -1)
2. Loop until low <= high
3. Check if middle value(low + high/2) < or > target, assign low and high to the respective left or right side

```python
# must be sorted first!
numbers = [2, 4, 6, 8, 10, 12]

def binary_search(numbers, target): 
    left = 0
    right = len(numbers) - 1
    
    while left <= right: 
        mid = (right + left) // 2
        midValue = numbers[mid]
        # found it
        if target == midValue: 
            return mid
        # exists in left subarray
        if target <= midValue:
            right = mid - 1
        # exists in right subarray
        else: 
            left = mid + 1
    # not found
    return -1

print(binary_search(numbers, 12))
```

## Breadth First Search ( BFS )
Visits all nodes level by level starting with a root node.

1. Enqueue the source node and mark as visited. 
2. Dequeue the node, enqueue the neighbor nodes, mark the neighbor nodes as visited. 
3. Continue level by level repeating the steps above until all nodes have been visited. 

**Use Cases**

Works for both graphs and trees. 
- In a graph, loops are possible so must use a visited array

Detect cycle in a graph, find shortest path in an unweighted graph.

```python

from collections import deque


def breadthFirstSearch(graph: dict[str, list[str]], start_node: str): 
    visited = set({start_node}) 
    queue = deque([start_node]) 
    result = []
    
    while queue: 
        current_node = queue.popleft()
        result.append(current_node)
        
        for neighbor in graph[current_node]:
            if(neighbor not in visited):
                queue.append(neighbor)
                visited.add(neighbor)
        
    return result

input =  {
        "A": ["B", "C"],
        "B": ["A", "D", "E"],
        "C": ["A", "F"],
        "D": ["B"],
        "E": ["B", "F"],
        "F": ["C", "E"],
    }

print(breadthFirstSearch(input, "A")) # ['A', 'B', 'C', 'D', 'E', 'F']
```

## Depth First Search ( DFS )
Traverse all adjacent vertices one by one. Traverse all vertices reachable by the adjacent vertex. Once it reaches a vertex with no unvisted neighbors it backtracks to previous vertex to explore unvisited paths. 

1. Check if node has been visited
2. If not visited then add mark as visited and add to result
3. Recursively visit all neighbor nodes
    - Each recursive call puts node on the call stack which backtracks through previously visited nodes

The alternative method to using recursion to follow the same logic for bfs but using a stack instead of a queue. 


```python

def depth_first_search(graph: dict[str, list[str]], node: str, visited=None, result=None): 
    if visited is None: 
        visited = set()
    if result is None: 
        result = []
    
    if node not in visited: 
        result.append(node)
        visited.add(node)
        for neighbor in graph[node]:
            depth_first_search(graph, neighbor, visited, result)
            
    return result
 

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': []
}

print(depth_first_search(graph, 'A')) # ['A', 'B', 'D', 'E', 'C', 'F']
```
