
## Breadth First Search
Visits all nodes level by level starting with a root node.

1. Enqueue the source node and mark as visited. 
2. Dequeue the node, enqueue the neighbor nodes, mark the neighbor nodes as visited. 
3. Continue level by level repeating the steps above until all nodes have been visited. 

**Use Cases**

Works for both graphs and trees. 

Detect cycle in a graph, find shortest path in an unweighted graph.

**Example**

Input: adj[][] = [[1, 2], [0, 2], [0, 1, 3, 4], [2], [2]]



- Output: [0, 1, 2, 3, 4] 
- Explanation: Starting from 0, the BFS traversal proceeds as follows:
- Visit 0 - Print: 0
- Visit 1 (neighbor of 0) - Print: 1
- Visit 2 (next neighbor of 0) - Print: 2
- Visit 3 (first neighbor of 2 that hasn't been visited yet) - Print: 3
- Visit 4 (next neighbor of 2) - Print: 4

Push current node, check which neighbors have not been visited, push its unvisited neighbors in the queue 





```python
from collections import deque


def breadthFirstSearch(graph: dict[str, list[str]], startNode: str): 
    result = []
    queue = deque([startNode])
    visited = set(startNode)
    
    while queue: 
        currentNode = queue.popleft()
        result.append(currentNode)
        for neighbor in graph[currentNode]: 
            if(neighbor not in visited): 
                queue.append(neighbor)
                visited.add(neighbor)
    return result; 
    
input =  {
        "A": ["B", "C"],
        "B": ["A", "D", "E"],
        "C": ["A", "F"],
        "D": ["B"],
        "E": ["B", "F"],
        "F": ["C", "E"],
    }

print(breadthFirstSearch(input, "A"))
```
